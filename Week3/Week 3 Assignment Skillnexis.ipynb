```python
#Replacing missing values in column Calories with mean value
df['Calories'] = df['Calories'].fillna(df['Calories'].mean(), inplace=True)
```

```python
df['Calories'].isnull().sum() # Count missing values
```

```python
#changing the data type of the column Date to datetime format
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')
```

```python
df['Date'] = pd.to_datetime(df['Date'].astype(str).str.strip('\'"'))
```

Inserting two New Columns - 'Calories_per_min' and 'pulse_range'

```python
df['Calories_per_min'] = df['Calories'] / df['Duration']
```

```python
df['Pulse_Range'] = df['Maxpulse'] - df['Pulse']
```
